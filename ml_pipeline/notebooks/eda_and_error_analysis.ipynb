# NephroScan AI: Exploratory Data Analysis & Error Diagnostics

This notebook analyzes the B-mode kidney ultrasound dataset distribution, image dimensions, intensity histograms, and Out-of-Distribution (OOD) test cases.

```python
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from PIL import Image

sys.path.append('..')
import config

print('Loaded ML Pipeline Config')
print('Train Path:', config.TRAIN_DIR)
print('Model Path:', config.BEST_MODEL_PATH)
```

### 1. Dataset Class Balance

```python
train_stone = len(list((config.TRAIN_DIR / 'stone').glob('*')))
train_normal = len(list((config.TRAIN_DIR / 'no_stone').glob('*')))

plt.figure(figsize=(6, 4))
sns.barplot(x=['Stone', 'No Stone'], y=[train_stone, train_normal], palette=['#0d9488', '#3b82f6'])
plt.title('Training Split Class Balance')
plt.ylabel('Count')
plt.show()
```
